```python
import pandas as pd
import numpy as np
import pickle
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv('start_df.csv')

X_test_full = df.drop(['MedHouseVal', 'is_capped'], axis=1)
y_test_full = df['MedHouseVal']

with open('classifier_model.pkl', 'rb') as f:
    clf = pickle.load(f)

with open('regressor_model.pkl', 'rb') as f:
    reg = pickle.load(f)
```

```python
y_capped_pred = clf.predict(X_test_full)
y_cascade_pred = np.full(len(X_test_full), 5.0)
```

```python
normal_mask = y_capped_pred == 0
if normal_mask.any():
    y_cascade_pred[normal_mask] = reg.predict(X_test_full[normal_mask])
```

```python
print(f"  MAE:  {mean_absolute_error(y_test_full, y_cascade_pred):.3f}")
print(f"  RMSE: {np.sqrt(mean_squared_error(y_test_full, y_cascade_pred)):.3f}")
print(f"  R2:   {r2_score(y_test_full, y_cascade_pred):.3f}")
```
