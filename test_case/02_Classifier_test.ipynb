```python
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb
import catboost as cat
import lightgbm as gbm

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import classification_report

import pandas as pd
import numpy as np

import pickle
```

```python
df = pd.read_csv('start_df.csv')

x = df.drop(['is_capped', 'MedHouseVal'], axis=1)
y = df['is_capped']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0, stratify=y)

param_set1 = [
    {'n_estimators': 100, 'max_depth': 6},  
    {'n_estimators': 150, 'max_depth': 8}, 
    {'n_estimators': 200, 'max_depth': 4}
]

param_set2 = [
    {'n_estimators': 100, 'max_depth': 6, 'learning_rate': 0.1},  
    {'n_estimators': 150, 'max_depth': 8, 'learning_rate': 0.05}, 
    {'n_estimators': 200, 'max_depth': 4, 'learning_rate': 0.2}
]

 
```

```python
for params in param_set2:
    clf = xgb.XGBClassifier(**params)
    clf.fit(x_train, y_train)
    y_pred = clf.predict(x_test)
    print(classification_report(y_test, y_pred))
```

```python
for params in param_set2:
    clf = cat.CatBoostClassifier(**params, verbose=False)
    clf.fit(x_train, y_train)
    y_pred = clf.predict(x_test)
    print(classification_report(y_test, y_pred))
```

## ИТОГ
Лучшим по всем показателям оказался XGBoost. В качестве классификатора выбираем его. Он нашел наибольший процент дорогих домов, и среди них меньше всего ошибся. В данном случае нам подходят параметры:  
**{'n_estimators': 200, 'max_depth': 4, 'learning_rate': 0.2}**  
При которых имеем precision, recall, f1 в случае дорогих домов:  
**XGBoost  - 0.86  	0.63	0.73**

```python
clf = xgb.XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.2)
clf.fit(x_train, y_train)
y_pred = clf.predict(x_test)
print(classification_report(y_test, y_pred))

with open('classifier_model.pkl', 'wb') as f:
    pickle.dump(clf, f)
```
